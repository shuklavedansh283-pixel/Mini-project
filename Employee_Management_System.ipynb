```python
# ===================================================
# STEP 1: Importing Libraries
# ===================================================
# NumPy helps us create repeatable random employee numbers.
import numpy as np
# Pandas helps us store and work with employee rows and columns.
import pandas as pd
```

```python
# ===================================================
# STEP 2: Creating Employee Dataset
# ===================================================
# A fixed seed makes the generated numbers the same every time.
np.random.seed(42)
# These lists provide the employee names, departments, and cities.
names = ['Aarav Sharma', 'Mia Patel', 'Noah Kim', 'Zara Khan', 'Liam Chen', 'Isha Rao', 'Ethan Lee', 'Anaya Shah', 'Lucas Wong', 'Riya Das', 'Oliver Scott', 'Sana Ali', 'Mason Reed', 'Diya Nair', 'Leo Martin', 'Aisha Bose', 'Jack Turner', 'Nina Roy', 'Arjun Mehta', 'Emma Clark', 'Kabir Joshi', 'Sara Thomas', 'Ryan Cooper', 'Maya Singh', 'Aiden Murphy', 'Leah Wilson', 'Omar Hassan', 'Chloe Baker', 'Eli Foster', 'Priya Kapoor', 'Henry Adams', 'Amira Yusuf', 'Caleb Young', 'Tara Menon', 'Isaac Green', 'Nora Evans', 'Rohan Verma', 'Lily Morgan', 'Adam Brooks', 'Sofia Rivera', 'Dev Shah', 'Grace Hall', 'Samir Khan', 'Ella Wright', 'Jay Patel', 'Mila Turner', 'Kiran Das', 'Luca Martin', 'Anika Roy', 'Theo Clark']
departments = ['HR', 'IT', 'Sales', 'Finance', 'IT', 'Sales', 'HR', 'Finance', 'IT', 'HR', 'Sales', 'Finance', 'IT', 'Sales', 'HR', 'Finance', 'IT', 'Sales', 'Finance', 'HR', 'IT', 'Sales', 'Finance', 'HR', 'IT', 'Sales', 'Finance', 'HR', 'IT', 'Sales', 'Finance', 'HR', 'IT', 'Sales', 'Finance', 'HR', 'IT', 'Sales', 'Finance', 'HR', 'IT', 'Sales', 'Finance', 'HR', 'IT', 'Sales', 'Finance', 'HR', 'IT', 'Sales']
cities = ['New York', 'Chicago', 'Austin', 'Seattle', 'Boston', 'Denver', 'Miami', 'Dallas', 'San Jose', 'Atlanta', 'Portland', 'Phoenix', 'Chicago', 'Seattle', 'Austin', 'Boston', 'Denver', 'Miami', 'Dallas', 'New York', 'San Jose', 'Atlanta', 'Portland', 'Phoenix', 'New York', 'Chicago', 'Austin', 'Seattle', 'Boston', 'Denver', 'Miami', 'Dallas', 'San Jose', 'Atlanta', 'Portland', 'Phoenix', 'Chicago', 'Seattle', 'Austin', 'Boston', 'Denver', 'Miami', 'Dallas', 'New York', 'San Jose', 'Atlanta', 'Portland', 'Phoenix', 'Chicago', 'Seattle']
# These generated values stay within the requested ranges.
ages = np.random.randint(22, 56, size=50)
salaries = np.random.randint(25000, 120001, size=50)
# Experience is capped so it never exceeds the employee's working-age years.
experience = np.random.randint(1, 26, size=50)
experience = np.minimum(experience, ages - 18)
# Two missing city values let us practice checking and filling missing data.
cities[3] = np.nan
cities[17] = np.nan
# Employee IDs start at 101 and increase by one for each employee.
emp_ids = np.arange(101, 151)
# Joining dates are stored as Pandas dates, 45 days apart.
joining_dates = pd.date_range(start='2021-01-15', periods=50, freq='45D')
# The lists and generated arrays are combined into one DataFrame.
df = pd.DataFrame({'Emp_ID': emp_ids, 'Name': names, 'Department': departments, 'Age': ages, 'Salary': salaries, 'Experience': experience, 'City': cities, 'Joining_Date': joining_dates})
# Show a few rows to confirm that the dataset was created.
print('Employee dataset created:')
print(df.head())
print('Number of employees:', len(df))
```

```python
# ===================================================
# STEP 3: Exploring Data
# ===================================================
# head() shows the first five employee rows.
print('First five employees:')
print(df.head())
# tail() shows the last five employee rows.
print('\nLast five employees:')
print(df.tail())
# shape reports the number of rows and columns.
print('\nDataset shape (rows, columns):', df.shape)
# columns lists all the column names.
print('\nColumn names:', df.columns.tolist())
# info() displays column types and non-missing value counts.
print('\nColumn information:')
df.info()
# describe() gives basic statistics for numeric columns.
print('\nNumeric summary:')
print(df.describe())
```

```python
# ===================================================
# STEP 4: Adding an Employee
# ===================================================
# This function adds one employee when the ID is not already in use.
def add_employee(emp_id, emp_name, department, age, salary, experience, city, joining_date):
    global df
    if emp_id in df['Emp_ID'].values:
        print('Employee ID already exists:', emp_id)
        return
    new_employee = {'Emp_ID': emp_id, 'Name': emp_name, 'Department': department, 'Age': age, 'Salary': salary, 'Experience': experience, 'City': city, 'Joining_Date': pd.to_datetime(joining_date)}
    df = pd.concat([df, pd.DataFrame([new_employee])], ignore_index=True)
    print('Employee added successfully.')
# Add a sample employee to demonstrate the function.
add_employee(151, 'Avery Brooks', 'IT', 29, 76000, 6, 'Denver', '2024-02-01')
# Display the DataFrame so the added row is visible.
print(df)
```

```python
# ===================================================
# STEP 5: Updating Employee
# ===================================================
# This function changes the salary, department, or city for one ID.
def update_employee(emp_id, new_salary=None, new_department=None, new_city=None):
    global df
    employee_rows = df['Emp_ID'] == emp_id
    if not employee_rows.any():
        print('Employee ID not found:', emp_id)
        return
    if new_salary is not None:
        df.loc[employee_rows, 'Salary'] = new_salary
    if new_department is not None:
        df.loc[employee_rows, 'Department'] = new_department
    if new_city is not None:
        df.loc[employee_rows, 'City'] = new_city
    print('Employee details updated for ID:', emp_id)
# Update the salary, department, and city of a sample employee.
update_employee(104, new_salary=84000, new_department='IT', new_city='Boston')
# Display the DataFrame so the changes are visible.
print(df)
```

```python
# ===================================================
# STEP 6: Deleting Employee
# ===================================================
# This function removes one employee row using the employee ID.
def delete_employee(emp_id):
    global df
    employee_rows = df['Emp_ID'] == emp_id
    if not employee_rows.any():
        print('Employee ID not found:', emp_id)
        return
    df = df.loc[~employee_rows].reset_index(drop=True)
    print('Employee deleted successfully:', emp_id)
# Delete one sample employee to demonstrate the function.
delete_employee(105)
# Display the DataFrame so the deletion is visible.
print(df)
```

```python
# ===================================================
# STEP 7: Searching Employee
# ===================================================
# This function searches by employee ID or by a name, ignoring letter case.
def search_employee(search_value, search_by='id'):
    if search_by == 'id':
        employee_rows = df.loc[df['Emp_ID'] == search_value]
    elif search_by == 'name':
        employee_rows = df.loc[df['Name'].str.lower() == str(search_value).lower()]
    else:
        print("Choose 'id' or 'name' when searching.")
        return
    if employee_rows.empty:
        print('No employee found for', search_by, ':', search_value)
    else:
        print('Employee search results:')
        print(employee_rows)
# Search once by ID and once by name.
search_employee(101, search_by='id')
search_employee('Aarav Sharma', search_by='name')
# Search for a missing ID to see the not-found message.
search_employee(999, search_by='id')
```

```python
# ===================================================
# STEP 8: Filtering Data
# ===================================================
# Select employees who work in the IT department.
it_employees = df.loc[df['Department'] == 'IT']
print('Employees in the IT department:')
print(it_employees)
# Select employees whose salary is above 80000.
high_salary_employees = df.loc[df['Salary'] > 80000]
print('\nEmployees earning more than 80000:')
print(high_salary_employees)
# Select employees who live in Boston.
boston_employees = df.loc[df['City'] == 'Boston']
print('\nEmployees in Boston:')
print(boston_employees)
```

```python
# ===================================================
# STEP 9: Sorting Data
# ===================================================
# Sort employees from the highest salary to the lowest salary.
salary_sorted = df.sort_values(by='Salary', ascending=False)
print('Employees sorted by salary (highest to lowest):')
print(salary_sorted)
# Sort employees by experience from the most experience to the least.
experience_sorted = df.sort_values(by='Experience', ascending=False)
print('\nEmployees sorted by experience:')
print(experience_sorted)
```

```python
# ===================================================
# STEP 10: Group Analysis
# ===================================================
# groupby() collects employees into departments for comparison.
department_average_salary = df.groupby('Department')['Salary'].mean()
print('Average salary by department:')
print(department_average_salary)
# Count the employees in every department.
department_employee_count = df.groupby('Department')['Emp_ID'].count()
print('\nEmployee count by department:')
print(department_employee_count)
# Find the highest salary in every department.
department_max_salary = df.groupby('Department')['Salary'].max()
print('\nMaximum salary by department:')
print(department_max_salary)
```

```python
# ===================================================
# STEP 11: Highest and Lowest Salary
# ===================================================
# idxmax() finds the row number of the highest salary.
highest_paid_row = df['Salary'].idxmax()
highest_paid_employee = df.loc[highest_paid_row]
print('Full details of the highest-paid employee:')
print(highest_paid_employee)
# idxmin() finds the row number of the lowest salary.
lowest_paid_row = df['Salary'].idxmin()
lowest_paid_employee = df.loc[lowest_paid_row]
print('\nFull details of the lowest-paid employee:')
print(lowest_paid_employee)
```

```python
# ===================================================
# STEP 12: NumPy Statistics
# ===================================================
# Convert the Salary column into a NumPy array for calculations.
salary_array = df['Salary'].to_numpy()
# NumPy calculates common summary statistics for salaries.
avg_salary = np.mean(salary_array)
median_salary = np.median(salary_array)
max_salary = np.max(salary_array)
min_salary = np.min(salary_array)
salary_std = np.std(salary_array)
# Print each result with a clear label.
print('Average Salary:', avg_salary)
print('Median Salary:', median_salary)
print('Maximum Salary:', max_salary)
print('Minimum Salary:', min_salary)
print('Salary Standard Deviation:', salary_std)
```

```python
# ===================================================
# STEP 13: Creating Bonus and Total Salary
# ===================================================
# Employees with over 10 years get a 10 percent bonus.
bonus_over_10 = np.where(df['Experience'] > 10, df['Salary'] * 0.10, 0)
# Employees with 5 through 10 years get a 5 percent bonus.
bonus_5_to_10 = np.where((df['Experience'] >= 5) & (df['Experience'] <= 10), df['Salary'] * 0.05, 0)
# Employees with less than 5 years get a 2 percent bonus.
bonus_under_5 = np.where(df['Experience'] < 5, df['Salary'] * 0.02, 0)
# Add the applicable bonus amounts into one Bonus column.
df['Bonus'] = bonus_over_10 + bonus_5_to_10 + bonus_under_5
# Add salary and bonus to make each employee's total salary.
df['Total_Salary'] = df['Salary'] + df['Bonus']
# Display the salary columns to check the calculations.
print('Salary, bonus, and total salary:')
print(df[['Emp_ID', 'Name', 'Experience', 'Salary', 'Bonus', 'Total_Salary']])
```

```python
# ===================================================
# STEP 14: Checking and Handling Missing Values
# ===================================================
# Count missing values in every column before filling them.
print('Missing values before filling:')
print(df.isna().sum())
# Fill missing city names with the simple label Unknown.
df['City'] = df['City'].fillna('Unknown')
# Check again to confirm there are no missing values left.
print('\nMissing values after filling:')
print(df.isna().sum())
# Display the final rows after the missing values are handled.
print('\nUpdated employee data:')
print(df)
```

```python
# ===================================================
# STEP 15: Saving Final Data to CSV
# ===================================================
# Save the final DataFrame beside this notebook.
csv_file = 'employee_data.csv'
df.to_csv(csv_file, index=False)
print('Employee data saved to:', csv_file)
# Read the saved file back into a new DataFrame.
saved_df = pd.read_csv(csv_file, parse_dates=['Joining_Date'])
# Compare the row and column counts to confirm that the file was saved.
print('Original DataFrame shape:', df.shape)
print('Data read back from CSV shape:', saved_df.shape)
print('Number of saved employee records:', len(saved_df))
print('First five saved records:')
print(saved_df.head())
print('CSV save check passed:', saved_df.shape == df.shape)
```
